```python
appearances = pd.read_csv('../data/raw/appearances.csv')
print(appearances.shape)
appearances.head()
```

```python
valuations = pd.read_csv('../data/raw/player_valuations.csv')
print(valuations.shape)
valuations.head()
```

```python
import requests

API_KEY = "2a75dfb3bfa949809e2e2512c497316f"
BASE_URL = "https://api.football-data.org/v4"

headers = {
    "X-Auth-Token": API_KEY}

response = requests.get(f"{BASE_URL}/competitions/PL/standings", headers=headers)
print(response.status_code)
data = response.json()
data
```

```python
table = data['standings'][0]['table']

standings_rows = []
for row in table:
    standings_rows.append({
        'position': row['position'],
        'team': row['team']['name'],
        'played': row['playedGames'],
        'won': row['won'],
        'draw': row['draw'],
        'lost': row['lost'],
        'points': row['points'],
        'goals_for': row['goalsFor'],
        'goals_against': row['goalsAgainst'],
        'goal_difference': row['goalDifference']
    })

standings_df = pd.DataFrame(standings_rows)
standings_df
```

```python
import os

os.makedirs('../data/api_raw', exist_ok=True)
standings_df.to_csv('../data/api_raw/pl_standings_2026.csv', index=False)
```

```python
os.listdir('../data/api_raw')
```

```python
player_stats = appearances.groupby('player_id').agg(
    total_games=('appearance_id', 'count'),
    total_goals=('goals', 'sum'),
    total_assists=('assists', 'sum'),
    total_minutes=('minutes_played', 'sum'),
    total_yellow_cards=('yellow_cards', 'sum'),
    total_red_cards=('red_cards', 'sum')
).reset_index()

player_stats.head()
```

```python
latest_valuations = valuations.sort_values('date').groupby('player_id').tail(1)
latest_valuations = latest_valuations[['player_id', 'date', 'market_value_in_eur']]
latest_valuations = latest_valuations.rename(columns={'date': 'valuation_date'})

latest_valuations.head()
```

```python
player_master = players.merge(player_stats, on='player_id', how='left')
player_master = player_master.merge(latest_valuations, on='player_id', how='left')

print(player_master.shape)
player_master.head()
```

```python
os.makedirs('../data/processed', exist_ok=True)
player_master.to_csv('../data/processed/player_master_raw.csv', index=False)
```

```python
missing = player_master.isna().sum().sort_values(ascending=False)
missing_pct = (missing / len(player_master) * 100).round(1)
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
```

```python
# Compare to understand the difference
player_master[['player_id', 'market_value_in_eur_x', 'market_value_in_eur_y', 'valuation_date']].sample(5)
```

```python
player_master = player_master.drop(columns=['market_value_in_eur_y'])
player_master = player_master.rename(columns={'market_value_in_eur_x': 'market_value_in_eur'})

player_master.columns.tolist()
```

```python
player_master['international_caps'] = player_master['international_caps'].fillna(0)
player_master['international_goals'] = player_master['international_goals'].fillna(0)

player_master_clean = player_master.dropna(subset=['total_games']).copy()

print(player_master.shape)
print(player_master_clean.shape)
```

```python
player_master_clean['foot'] = player_master_clean['foot'].fillna('Unknown')
player_master_clean['country_of_birth'] = player_master_clean['country_of_birth'].fillna('Unknown')
player_master_clean['height_in_cm'] = player_master_clean['height_in_cm'].fillna(
    player_master_clean['height_in_cm'].median()
)

print(player_master_clean[['foot', 'country_of_birth', 'height_in_cm']].isna().sum())
```

```python
player_master_clean = player_master_clean.dropna(subset=['market_value_in_eur'])
print(player_master_clean.shape)
```

```python
remaining_missing = player_master_clean.isna().sum()
remaining_missing[remaining_missing > 0]
```

```python
player_master_clean['first_name'] = player_master_clean['first_name'].fillna('Unknown')
player_master_clean['city_of_birth'] = player_master_clean['city_of_birth'].fillna('Unknown')
player_master_clean['country_of_citizenship'] = player_master_clean['country_of_citizenship'].fillna('Unknown')
player_master_clean['sub_position'] = player_master_clean['sub_position'].fillna('Unknown')
player_master_clean = player_master_clean.dropna(subset=['date_of_birth'])
player_master_clean = player_master_clean.dropna(subset=['current_club_name'])

print(player_master_clean.shape)
```

```python
player_master_clean[['height_in_cm', 'market_value_in_eur', 'total_goals', 'total_minutes']].describe()
```

```python
# suspicious short/tall entries
player_master_clean[player_master_clean['height_in_cm'] < 140][['name', 'height_in_cm', 'position']]
```

```python
median_height = player_master_clean['height_in_cm'].median()
player_master_clean.loc[player_master_clean['height_in_cm'] < 140, 'height_in_cm'] = median_height

player_master_clean['height_in_cm'].describe()
```

```python
player_master_clean.to_csv('../data/processed/player_master_clean.csv', index=False)
print("Saved:", player_master_clean.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs('../images', exist_ok=True)

fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(player_master_clean['market_value_in_eur'], bins=50, log_scale=True, ax=ax)
ax.set_title('Distribution of Player Market Values')
ax.set_xlabel('Market Value (EUR)')
ax.set_ylabel('Number of Players')
plt.tight_layout()
plt.savefig('../images/01_market_value_distribution.png', dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=player_master_clean, x='position', y='market_value_in_eur', ax=ax, color='lightgreen')
ax.set_yscale('log')
ax.set_title('Market Value by Position')
ax.set_xlabel('Position')
ax.set_ylabel('Market Value (EUR)')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig('../images/02_value_by_position.png', dpi=150)
plt.show()
```

```python
player_master_clean['date_of_birth'] = pd.to_datetime(player_master_clean['date_of_birth'])
player_master_clean['age'] = (pd.Timestamp('2026-09-18') - player_master_clean['date_of_birth']).dt.days // 365

fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(data=player_master_clean, x='age', y='market_value_in_eur', alpha=0.3, ax=ax, color='orange')
ax.set_yscale('log')
ax.set_title('Player Age vs. Market Value')
ax.set_xlabel('Age')
ax.set_ylabel('Market Value (EUR)')
plt.tight_layout()
plt.savefig('../images/03_age_vs_value.png', dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(data=player_master_clean, x='total_goals', y='total_assists', hue='position', alpha=0.5, ax=ax)
ax.set_title('Career Goals vs. Assists by Position')
ax.set_xlabel('Total Goals')
ax.set_ylabel('Total Assists')
plt.tight_layout()
plt.savefig('../images/04_goals_vs_assists.png', dpi=150)
plt.show()
```

```python
top_clubs = (player_master_clean.groupby('current_club_name')['market_value_in_eur']
             .mean().sort_values(ascending=False).head(10))

fig, ax = plt.subplots(figsize=(9, 5))
top_clubs.plot(kind='barh', ax=ax, color='brown')
ax.invert_yaxis()
ax.set_title('Top 10 Clubs by Average Player Market Value')
ax.set_xlabel('Average Market Value (EUR)')
ax.set_ylabel('Club')
plt.tight_layout()
plt.savefig('../images/05_top_clubs_by_value.png', dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
sns.barplot(data=player_master_clean, x='foot', y='market_value_in_eur', ax=ax, estimator='mean')
ax.set_title('Average Market Value by Preferred Foot')
ax.set_xlabel('Preferred Foot')
ax.set_ylabel('Average Market Value (EUR)')
plt.tight_layout()
plt.savefig('../images/06_value_by_foot.png', dpi=150)
plt.show()
```
